```python
import pandas as pd

data = {
    'Name':['A','B','C','D','E'],
    'Math':[78,67,90,56,88],
    'Science':[85,72,95,60,84],
    'English':[88,70,92,58,86]
}
df = pd.DataFrame(data)
print(df)
```

```python
marks = df[["Math","Science","English"]].values
average_marks = marks.mean(axis=1)
df["Average Marks"] = average_marks
print(df)
```

```python

def grade(marks):
    if marks >= 90:
       return 'A'
    elif marks >= 80:
       return 'B'
    elif marks >= 70:
       return 'C'
    elif marks >= 60:
       return 'D'
    else:
       return 'F'

df['Grade'] = df['Average Marks'].apply(grade)
df
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.bar(df['Name'], df['Average Marks'])
plt.xlabel('Name')
plt.ylabel('Average Marks')
plt.title('Average Marks of Students')
plt.show()
```

```python
df[["Math","Science","English"]].mean().plot(kind="bar")
plt.ylabel("Average Marks")
plt.title("Average Marks per Subjects")
plt.show()
```
