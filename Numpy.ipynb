```python
import numpy as np
students = np.array([
    [65, 90, 70], # student 1
    [60, 55, 70], # student 2
    [95, 92, 88], # student 3
    [40, 50, 45], # student 4
    [72, 68, 75], # student 5
])
students
```

```python
avg_marks = np.mean(students, axis=1)
print("Average Marks : " , avg_marks)
```

```python
topper= students[avg_marks.argmax()]
print("Topper : " , topper)
```

```python
topper= np.argmax(avg_marks)
print("Topper : " , topper)
```

```python
failed_students = students[avg_marks<50]
print("Failed Students : " , failed_students)
```

```python
failed_students_count = np.count_nonzero(avg_marks<50)
print("Failed Students Count : " , failes_students_count)
```

```python
#performance Report
for i, avg in enumerate(avg_marks):
    status= "pass" if avg >= 50 else "fail"
    print(f"Student {i+1} : avg= {avg:.2f} | {status}")
```
