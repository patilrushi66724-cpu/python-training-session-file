```python
import pandas as pd
data={
    'name':['Alice', 'Bob', 'Claire'],
    'age': [25, 30, 27],
    'city': ['New York', 'San Francisco', 'Los Angeles']
}
df = pd.DataFrame(data)
print(df)
```

```python
#add column
df['salary'] = [50000, 60000, 70000]
print(df)
```

```python
# Check if 'salary' column exists; if not, initialize it first
if 'salary' in df.columns:
    df['salary'] = df['salary'] + 10000
else:
    # Initialize with default values if it was dropped
    df['salary'] = [50000, 60000, 70000]
    df['salary'] = df['salary'] + 10000
df
```

```python
# Using errors='ignore' prevents a KeyError if the column is already dropped
df.drop('salary', axis=1, inplace=True, errors='ignore')
df
```

```python
# add the row with values for all 4 columns (name, age, city, salary)
df.loc[3] = ['Ram', 28, 'Delhi', 75000]
df.loc[4] = ['Shyam', 29, 'Mumbai', 80000]
df.loc[5] = ['Rosa', 30, 'London', 85000]
df
```

```python
# Handling missing data
df.fillna(0, inplace=True)
df
```

```python
# grouping and aggregation
df.groupby('city')['salary'].mean()
```

```python
# sorting data
df.sort_values(by='salary', ascending=False)
```

```python
# value count
df['city'].value_counts()
pd.crosstab(df['city'], df['age'])
```
