```python
class A:
  def showA(self):
    print("I am in class A")

class B(A):
  def showB(self):
    print("I am in class B")
class C(B):
  def showC(self):
    print("showC function")

obj = C()
obj.showA()
obj.showB()
obj.showC()
```

```python
class A:
  def showA(self):
    print("showA function")

class B:
  def showB(self):
    print("showB function")
class C(A,B):
  def showC(self):
    print("showC function")

obj =C()
obj.showA()
obj.showB()
obj.showC()
```
